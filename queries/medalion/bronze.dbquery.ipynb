```sql
SELECT
meet.meeting_key,
meet.meeting_name,
-- meet.meeting_official_name,
-- meet.country_code,
meet.country_name,
meet.location,
meet.date_start,
meet.date_end,
-- meet.is_cancelled,
-- meet.gmt_offset,
ses.session_key,
ses.session_name,
ses.session_type,
-- ses.circuit_key,
ses.circuit_short_name,
driv.driver_number,
driv.full_name,
driv.team_name,
driv.team_colour,
lap.lap_number,
lap.date_start,
lap.duration_sector_1,
lap.duration_sector_2,
lap.duration_sector_3,
lap.lap_duration,
lap.i1_speed,
lap.i2_speed,
lap.st_speed
-- lap.segments_sector_1,
-- lap.segments_sector_2,
-- lap.segments_sector_3
FROM
formula_1.bronze_meetings meet
JOIN formula_1.bronze_sessions ses ON meet.meeting_key = ses.meeting_key
JOIN formula_1.bronze_drivers driv ON meet.meeting_key = driv.meeting_key AND ses.session_key = driv.session_key
JOIN formula_1.bronze_laps lap ON meet.meeting_key = lap.meeting_key AND ses.session_key = lap.session_key AND driv.driver_number = lap.driver_number
WHERE meet.meeting_key IS NOT NULL
LIMIT 5;


SELECT
    meeting_name,
    -- session_name,
    -- session_type,
    full_name AS driver_name,
    COUNT(session_key) AS total_sessions_by_driver
FROM 
    formula_1.bronze_laps_by_meetings_sessions_drivers
GROUP BY
    meeting_name,
    -- session_name,
    -- session_type,
    driver_name
ORDER BY
    total_sessions_by_driver DESC;


SELECT * FROM formula_1.bronze_meetings ORDER BY date_start DESC LIMIT 5;

SELECT
*
FROM
formula_1.bronze_laps_by_meetings_sessions_drivers
WHERE meeting_key = 1285
;
```
