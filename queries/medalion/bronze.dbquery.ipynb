```sql
SELECT * FROM read_files('/Volumes/workspace/formula_1/formula_1/meetings', multiLine => true) LIMIT 5;
SELECT * FROM read_files('/Volumes/workspace/formula_1/formula_1/sessions', multiLine => true) LIMIT 5;
SELECT * FROM read_files('/Volumes/workspace/formula_1/formula_1/drivers', multiLine => true) LIMIT 5;
SELECT * FROM read_files('/Volumes/workspace/formula_1/formula_1/laps', multiLine => true) LIMIT 5;

SELECT "bronze_meetings" AS table_name, COUNT(*) AS count_rows FROM formula_1.bronze_meetings
UNION ALL
SELECT "bronze_sessions" AS table_name, COUNT(*) AS count_rows FROM formula_1.bronze_sessions
UNION ALL
SELECT "bronze_drivers" AS table_name, COUNT(*) AS count_rows FROM formula_1.bronze_drivers
UNION ALL
SELECT "bronze_laps" AS table_name, COUNT(*) AS count_rows FROM formula_1.bronze_laps;

-- SELECT
-- meet.meeting_key,
-- meet.meeting_name,
-- -- meet.meeting_official_name,
-- -- meet.country_code,
-- meet.country_name,
-- meet.location,
-- meet.date_start,
-- meet.date_end,
-- -- meet.is_cancelled,
-- -- meet.gmt_offset,
-- ses.session_key,
-- ses.session_name,
-- ses.session_type,
-- -- ses.circuit_key,
-- ses.circuit_short_name,
-- driv.driver_number,
-- driv.full_name,
-- driv.team_name,
-- driv.team_colour,
-- lap.lap_number,
-- lap.date_start,
-- lap.duration_sector_1,
-- lap.duration_sector_2,
-- lap.duration_sector_3,
-- lap.lap_duration,
-- lap.i1_speed,
-- lap.i2_speed,
-- lap.st_speed
-- -- lap.segments_sector_1,
-- -- lap.segments_sector_2,
-- -- lap.segments_sector_3
-- FROM
-- formula_1.bronze_meetings meet
-- JOIN formula_1.bronze_sessions ses ON meet.meeting_key = ses.meeting_key
-- JOIN formula_1.bronze_drivers driv ON meet.meeting_key = driv.meeting_key AND ses.session_key = driv.session_key
-- JOIN formula_1.bronze_laps lap ON meet.meeting_key = lap.meeting_key AND ses.session_key = lap.session_key AND driv.driver_number = lap.driver_number
-- WHERE meet.meeting_key IS NOT NULL
-- LIMIT 5;

-- Encontrar duplicados en todas las tablas bronze unificado
SELECT 
  'bronze_meetings' AS table_name,
  meeting_key AS key_value,
  NULL AS additional_key_1,
  NULL AS additional_key_2,
  NULL AS additional_key_3,
  COUNT(*) as duplicate_count
FROM formula_1.bronze_meetings
GROUP BY meeting_key
HAVING COUNT(*) > 1

UNION ALL

SELECT 
  'bronze_sessions' AS table_name,
  CAST(session_key AS STRING) AS key_value,
  CAST(meeting_key AS STRING) AS additional_key_1,
  NULL AS additional_key_2,
  NULL AS additional_key_3,
  COUNT(*) as duplicate_count
FROM formula_1.bronze_sessions
GROUP BY session_key, meeting_key
HAVING COUNT(*) > 1

UNION ALL

SELECT 
  'bronze_drivers' AS table_name,
  CAST(driver_number AS STRING) AS key_value,
  CAST(session_key AS STRING) AS additional_key_1,
  CAST(meeting_key AS STRING) AS additional_key_2,
  NULL AS additional_key_3,
  COUNT(*) as duplicate_count
FROM formula_1.bronze_drivers
GROUP BY driver_number, session_key, meeting_key
HAVING COUNT(*) > 1

UNION ALL

SELECT 
  'bronze_laps' AS table_name,
  CAST(driver_number AS STRING) AS key_value,
  CAST(lap_number AS STRING) AS additional_key_1,
  CAST(session_key AS STRING) AS additional_key_2,
  CAST(meeting_key AS STRING) AS additional_key_3,
  COUNT(*) as duplicate_count
FROM formula_1.bronze_laps
GROUP BY driver_number, lap_number, session_key, meeting_key
HAVING COUNT(*) > 1

ORDER BY table_name, duplicate_count DESC;



```
