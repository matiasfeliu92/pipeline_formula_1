```sql
SELECT * FROM formula_1.bronze_meetings LIMIT 5;
SELECT * FROM formula_1.bronze_sessions LIMIT 5;
SELECT * FROM formula_1.bronze_drivers LIMIT 5;
SELECT * FROM formula_1.bronze_laps LIMIT 5;

SELECT COUNT(*) FROM formula_1.bronze_sessions;
SELECT COUNT(*) FROM formula_1.bronze_drivers;

SELECT DISTINCT country_name FROM formula_1.bronze_meetings;

SELECT DISTINCT session_key FROM formula_1.bronze_sessions;

SELECT COUNT(*) FROM formula_1.bronze_laps_by_meetings_sessions_drivers


SELECT 
    meet.meeting_key,
    meet.meeting_name,
    ses.session_key, 
    driv.driver_number, 
    lap.lap_number, 
    COUNT(*) 
FROM formula_1.bronze_meetings meet
JOIN formula_1.bronze_sessions ses ON meet.meeting_key = ses.meeting_key
JOIN formula_1.bronze_drivers driv ON meet.meeting_key = driv.meeting_key AND ses.session_key = driv.session_key
JOIN formula_1.bronze_laps lap ON meet.meeting_key = lap.meeting_key AND ses.session_key = lap.session_key AND driv.driver_number = lap.driver_number
GROUP BY 1, 2, 3, 4, 5
HAVING COUNT(*) > 1;





-- Cantidad de filas en la versión actual
SELECT COUNT(*) AS current_count
FROM formula_1.bronze_meetings;

-- Cantidad de filas en la versión anterior (por ejemplo versión 5)
SELECT COUNT(*) AS previous_count
FROM formula_1.bronze_meetings VERSION AS OF 1;

SELECT * FROM formula_1.bronze_laps_by_meetings_sessions_drivers limit 1;


SELECT version
FROM (DESCRIBE HISTORY formula_1.bronze_meetings)
ORDER BY version DESC
LIMIT 1;
```
