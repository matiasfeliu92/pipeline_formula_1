```python
%sql
SELECT
  COUNT(*) AS total_rows,
  COUNT(DISTINCT circuit_key) AS unique_circuit_key,
  COUNT(DISTINCT circuit_short_name) AS unique_circuit_short_name,
  COUNT(DISTINCT circuit_type) AS unique_circuit_type,
  COUNT(DISTINCT country_code) AS unique_country_code,
  COUNT(DISTINCT country_key) AS unique_country_key,
  COUNT(DISTINCT country_name) AS unique_country_name,
  COUNT(DISTINCT location) AS unique_location,
  COUNT(DISTINCT meeting_key) AS unique_meeting_key,
  COUNT(DISTINCT meeting_name) AS unique_meeting_name,
  COUNT(DISTINCT meeting_official_name) AS unique_meeting_official_name,
  COUNT(DISTINCT year) AS unique_year,
  COUNT(DISTINCT month) AS unique_month,
  COUNT(DISTINCT is_cancelled) AS unique_is_cancelled,
  COUNT(DISTINCT gmt_offset) AS unique_gmt_offset,
  COUNT(DISTINCT file_metadata_path) AS unique_file_metadata_path,
  SUM(CASE WHEN circuit_key IS NULL THEN 1 ELSE 0 END) AS null_circuit_key,
  SUM(CASE WHEN circuit_short_name IS NULL THEN 1 ELSE 0 END) AS null_circuit_short_name,
  SUM(CASE WHEN circuit_type IS NULL THEN 1 ELSE 0 END) AS null_circuit_type,
  SUM(CASE WHEN country_code IS NULL THEN 1 ELSE 0 END) AS null_country_code,
  SUM(CASE WHEN country_key IS NULL THEN 1 ELSE 0 END) AS null_country_key,
  SUM(CASE WHEN country_name IS NULL THEN 1 ELSE 0 END) AS null_country_name,
  SUM(CASE WHEN location IS NULL THEN 1 ELSE 0 END) AS null_location,
  SUM(CASE WHEN meeting_key IS NULL THEN 1 ELSE 0 END) AS null_meeting_key,
  SUM(CASE WHEN meeting_name IS NULL THEN 1 ELSE 0 END) AS null_meeting_name,
  SUM(CASE WHEN meeting_official_name IS NULL THEN 1 ELSE 0 END) AS null_meeting_official_name,
  SUM(CASE WHEN year IS NULL THEN 1 ELSE 0 END) AS null_year,
  SUM(CASE WHEN month IS NULL THEN 1 ELSE 0 END) AS null_month,
  SUM(CASE WHEN is_cancelled IS NULL THEN 1 ELSE 0 END) AS null_is_cancelled,
  SUM(CASE WHEN gmt_offset IS NULL THEN 1 ELSE 0 END) AS null_gmt_offset,
  SUM(CASE WHEN file_metadata_path IS NULL THEN 1 ELSE 0 END) AS null_file_metadata_path
FROM
  formula_1.bronze_meetings;
```
